# Iter en python

Pensar en iter() en Python es muy fácil si te imaginas un dispensador de tickets de turno (como los del banco o de la carnicería).

1. La analogía del dispensador de tickets
Una Lista o Rebanada ([1, 2, 3]): Es una fila de personas paradas. Todas existen al mismo tiempo y ocupan espacio.

El iter(): Es el dispensador automático. No te entrega la fila completa, sino un mecanismo que te da un solo elemento a la vez cada vez que presionas el botón.

El next(): Es el acto de presionar el botón. Te entrega el elemento actual y el contador avanza a la siguiente posición de forma transparente.

```python
lista = ['a', 'b', 'c']
it = iter(lista) # Convertimos la lista en un objeto iterador
```

El objeto it guarda internamente un puntero que recuerda exactamente dónde se quedó.

Cada vez que ejecutas next(it), suceden dos cosas:

Te devuelve el valor en la posición actual.

Avanza el puntero interno un paso hacia adelante.

```python
print(next(it))  # Devuelve 'a' (el puntero avanza al siguiente)
print(next(it))  # Devuelve 'b' (el puntero avanza al siguiente)
print(next(it))  # Devuelve 'c' (el puntero llega al final)

print(next(it))  # ❌ ¡ERROR! Lanza la excepción StopIteration porque no quedan más elementos.
```

3. El truco del valor por defecto en next()
Para evitar que tu programa explote si no quedan más elementos en el iterador, la función next() te permite pasarle un segundo parámetro como "red de seguridad":

```python
# Sintentizador seguro:
valor = next(it, None)

# De tal manera que si no hay un elemento siguiente en lugar
# de rentorar un error retorna un "None"
```

4. ¿Por qué es una herramienta clave en Programación Competitiva?

Imagina que recibes esta entrada con la función sys.stdin.read().split():
['5', '3', '10', '8', '6', '11', '4', '1', '10']

```python
# Sin iter()
import sys
entrada = sys.stdin.read().split()

n = int(entrada[0])
x = list(map(int, entrada[1:n+1])) # ⚠️ 'entrada[1:n+1]' crea una NUEVA lista en memoria RAM
q = int(entrada[n+1])
# Tienes que ir sumando índices manualmente: n + 2, n + 3, etc.
```

```python
# con iter
it = iter(sys.stdin.read().split())

n = int(next(it))                   # Extrae el '5' y avanza el puntero
x = [int(next(it)) for _ in range(n)] # Consume automáticamente los siguientes 'n' elementos
q = int(next(it))                   # Extrae 'q' inmediatamente después
```

Ventajas
Ahorras memoria (Espacio $\mathcal{O}(1)$ adicional): No creas sublistas (entrada[1:n+1]) que dupliquen datos en la memoria RAM.Cero matemáticas con índices: No tienes que calcular si la posición de $q$ era n + 1 o n + 2. Solo llamas a next(it) y el iterador sabe exactamente qué número sigue.Súper rápido: Es una estructura nativa optimizada internamente en lenguaje C.
